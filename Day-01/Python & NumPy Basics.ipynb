```python
# Collection = Single "Variable" use to store multiple values 
#    List  = [] prdered and changeable. Duplicates OK
#    Set   = {} unordered and immutable, but Add/Remove OK. NO duplicates 
#    Tuple = () ordered and unchangeable. Duplicates Ok. FASTER

# List 

fruits = ['apple', 'ornage', 'banana', 'coconut']

print(fruits)

print(fruits[3])  # Indexing
fruits[0] = 'pineapple'
print(fruits)

print(fruits[0:3]) # slicing
print(fruits[::2]) # step of 2 



```

```python
# Method 

fruits.append('mango') # add a element to list in last posision 
print(fruits)

fruits.remove('ornage') # remove element for the list 
print(fruits)

fruits.insert(2,'grapes')  # adding eliment in a 2 index 
print(fruits)

fruits.sort() # short the order 
print(fruits)

fruits.reverse() # reverse the order 
print(fruits)

print(fruits.index('coconut')) # return the index of the coconut position on list 

print(fruits.count('banana'))  # count how many banana in the fruits list 

fruits.clear()  # to Clear a list 
print(fruits)
```

```python
# Set ->  {} unordered and immutable, but Add/Remove OK. NO duplicates 

fruits = {'apple', 'ornage', 'banana', 'coconut', 'coconut'}

print(fruits)

print(len(fruits)) # fint the len of any thing string,list,set etc

# method 
fruits.add('pineapple') # add element to set
print(fruits)

fruits.remove('banana') # remove element form set 
print(fruits)

fruits.pop() # remove random item form the list 

fruits.clear() # clear the set 
```

```python
# Tuple -> () ordered and unchangeable. Duplicates Ok. FASTER

fruits = ('apple', 'ornage', 'banana', 'coconut', 'coconut')

print(fruits)

# Method 

print(fruits.index('banana'))

fruits.count('coconut')
```

```python
# Global keyword ka use 
# Agar app function ke andar global variable ko change karna chahte ho: 

z = 5

def update():
    global z
    z = 10 

print(update())
print(z)
```

```python
# Nonlocal Scope 
# jab nested function (function ke andar function) ho 
# Aur inner function ko outer function ke variable ko change karna ho 

def outer():
    a = 50 

    def inner():
        nonlocal a
        a = 100

    inner()
    print(a)

outer()

# a outer function ka variable hai 
# inner function normally change nahi kar sakta 
# nonlocal bolta hai : 'Bhai outer function wala variable use karo'
```

```python
# Classes and Object 
# Class ek blueprint/design/plan hota hai 
# object class se bani hui real cheez , jo actually exist karta hai 

class Student:

    def __init__(self,name,age):
        self.name = name
        self.age = age

    def introduce(self):
        print('Mera naam', self.name, 'hai aur meri age',self.age, 'hai')


# Object create karna 
s1 = Student('Naidu', 20 )
s2 = Student('jerry', 30)

s1.introduce()
s2.introduce()


# Student -> Class
# s1 aut s2 -> Objects
# name,age -> properties (Attributes)
# introduce() -> Method(Funcion)
```

```python
# Inheritance 
# Socho:
# Parent -> Hieght, Eye color
# Child -> Same height, same eye + apna extra qualities 

class Animal:
    def speak(self):
        print('Animal bol raha hai')

class Dog(Animal):
    def bark(self):
        print('Dog bhook raha hai')


# Object create 
d = Dog()

d.bark() # Child class method 
d.speak() # Parent class method
```

```python
# Polymorphism 
# same function/ method, lekin different behavior.(ek hi chizz,alag-alag situation mein alag kaam karta hai)

# Python mein polymorphism mainly 2 ways se hota hai 
# 1.Method Overriding (inheritance ka saath)
# parent aur child class main same method name, lekin differnet implementation

class Animal:
    def sound(self):
        print('Animal sound')

class Dog(Animal):
    def sound(self):
        print('Dog bhonkta hai')

class Cat(Animal):
    def sound(self):
        print('Cat meow karti hai')


# object create
a = Animal()
d = Dog()
c = Cat()


a.sound()
d.sound()
c.sound()

# sound() method same hai 
# Behavior different hai -> this is Polymorphism 
```

```python
# 2.Method Overloading 
# python mein traditional overloading nai hota, lekin default arguments se achive kar sakte hain.as_integer_ratio

class Math:
    def add(self,a,b=0,c=0):
        return a+b+c
    
m = Math()

print(m.add(2))
print(m.add(2,3))
print(m.add(2,3,4))

# In Short Polymorphism ek Opp concept hai jisme same method ya function different forms mein behvae karta hai 
```

```python
# Encapsulation 
# Data ka hide karna + controlled access dena = Encapsulation 
# Python mein encapsulation acess modifiers ke through acheve hot hai:
# 1. Public Member (Directly accesible)

class Student:
    def __init__(self,name,):
        self.name = name  # public 


s = Student('Naidu')
print(s.name)
        
```

```python
# 2. Protected Members(_)
# Indirectly access 
# Child class ke liya safe

class Student:
    def __init__(self):
        self._college = 'ABC College'

s = Student()
print(s._college) # allowed, but not recommended 
```

```python
# 3. Private Members(_)
# Direct access allowed nahi 
# Strong data hiding 

class Student:
    def __init__(self):
        self.__fees = 5000

    def show_fees(self):
        print(self.__fees)


s = Student()

s.show_fees()
# s.__fees -> Error 
```

```python
## Gette aur Setter Methods 

# Private data ko safely access/update karne ke liya use hota hain.as_integer_ratio

class Student:
    def __init__(self):
        self.__marks = 0

    def set_marks(self,m):
        if m >= 0:
            self.__marks = m 

    def get_marks(self):
        return self.__marks
    

s = Student()

s.set_marks(85)
print(s.get_marks())

# Encapsulation ek Opp concept hai jisme data aur methods ko ek unit mein bind kiya jata hai aur data ko direct access se hid kiya jata hai.
```

```python
# Try-Except Blocks (Error handling ke liya use hota hai)

# try -> Yahan wo code likhte hain jisme error aa sakta hai 
# except -> agar error aaye, to yeh block execute hota hai 

try:
    a = 10 
    b = 0
    print(a/b)
except:
    print('Error aaya! Division by 0 allowed nahi hai.')
```

```python
# Specific Error Handle Karna (Best Practice)

try:
    x = int(input('Number daalo: '))
    print(10 / x)
except ZeroDivisionError:
    print('Zero se divide nai kar sakte!')
except ValueError:
    print('Srif number daalo!')

# Explanation:
# Agar user 0 daale -> ZeroDivisionError
# Afar user tst daale -> ValueError
# har error ka alag message 
```

```python
# else Block 
# else tab chalta hai jab error nahi aata 

try: 
    num = int(input('Number: '))
    print(100/ num)
except ZeroDivisionError:
    print('Zero allowed nahi')
else:
    print('Calculation successfull!')
```

```python
# Finally Block 
# Finally hamesha chalega, error aaya ya na aaye 

try:
    f = open('data.txt')
    print(f.read())
except:
    print('file nai mila')
finally:
    print('Program end ho raha hai')
```
