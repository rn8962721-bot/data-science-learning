#### Numpy Boradcasting

```python
# Numpy Boradcasting 
# Broadcasting Numpy ka ek smart feature hai jisme different shapes ke arrays ke beech arithmetic operations bina loop ke ho jaate hain.

import numpy as np 

arr = np.array([1,2,3])

print(arr + 10)

# 10 ko Numpy ne internally [10,10,10] maan liya
# har element ke sath add ho gaya 
```

#### Broadcasting Visualization


```python
# Bordcasting with 2D Array 

mat = np.array([[1,2,3],
                [4,5,6]])

print(mat  + 10)

# 10 har element me add ho gaya
# no loop no extra code 
```

```python
# Boradcasting : Row ke saath Operation

mat = np.array([[1,2,3],
               [4,5,6]])

row = np.array([10,20,30]) 

print(mat + row)

# row ka shape(3,)
# mat ka shape(2,3)
# Numpy ne row ko har row ke liya repeat kar diya 
```

```python
# Boradcasting: Column ke saath Operation

col = np.array([[10],
                [20]])

print(mat + col)

# Columns har row me add ho gaya 
```

```python
# Broadcasting Rule :
# Rule 1: Dimensions right se compare hote hai 
# Rule 2: Do dimensions compatible hote hain agar:
#           Dono equal ho - OR - Ek dimension 1 ho 
```

#### Handling missing data 

```python
# Missing Data kya hota hai?
## Missing values usually hoti hain:
#    - NaN
#    - None
#    - Empty cells
# Real Datasets me almost hamesha hota hain.
```

```python
# Missing Values Detect Karna 

df.isnull()

# Count check:

df.isnull().sum()  # Humesha pehle yeh chek karo.
```

```python
# Missing Data kyun Dangerous hai?
# -> ML modesl NaN accept nai karte 
# -> Mean / median galat ho jata hai 
# -> Bias aaa jata hai 

# isliya Missing value Handling mandatory hai.
```

```python
# Drop VS Fill (Decision Rule)
# -> Missing bahut zayada ho -> drop
# -> Missing kam ho -> fill
# -> Target column me missing -> careful
```

```python
# Row / Columns Drop karna 
# Row drop (missing vali row hata do) 

df.dropna()
```

```python
# Column drop

df.dropna(axis=1)

# Column tab drop karo jab > 40-50% data missing ho.
```

```python
## Fill Missing Value (Most Used)

# -> Numeric column -> Mean

df['Salary'].fillna(df['Salary'].mean(), inplace=True)
```

```python
# -> Numeric column -> Median (Better for Outliers)

df['Age'].fillna(df['Age'].median(), inplace=True)

### Rule:
# - Outliers ho -> Median
# - Normal distributin -> mean
```

```python
# Categorical Column -> Mode

df['City'].fillna(df['City'].mode(), inplace=True)
```

```python
# Forward Fill/ Backward Fill (Time Series)

df.ffill(inplace=True)
```

```python
# Backward Fill 

df.bfill(inplace=True)

# Mostly time-series data me use hota hai.
```

```python
## Replace Missing with constant 

df['Experience'].fillna(0, inplace=True)
```

```python
# Real Data Scientist Strategy 
#  Column Type ->  Best Approach 
#  Numeric     ->  Mean/Median
#  Categorical ->  Mode
#  Time Series ->  Forward fill / Backward fill
#  >50% missing -> Drop column
```

#### Data aggregation (groupby)

```python
# GroupBy ka simple matplab
# -> Data ko groups me todna + summary nikalna 

data = {
    "Department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
    "Employee": ["Amit", "Neha", "Rahul", "Priya", "Rohit", "Anjali"],
    "Salary": [60000, 65000, 50000, 52000, 45000, 48000],
    "Age": [30, 28, 35, 32, 26, 29]
}

df = pd.DataFrame(data)
```

```python
# Basic GroupBy(Most Common)

# 1. Department wise average salary  
df.groupby('Department')['Salary'].mean()
```

```python
# Multiple Aggrefations 

df.groupby('Department')['Salary'].agg(['mean','max','min'])
```

```python
# Multiple Columns Aggregation 

df.groupby('Department').agg({
    'Salary' : 'mean',
    'Age' : 'median'
})
```

```python
# GroupBy + Filtering (Very Important)

# -> Average salary > 50000 wale departments 

df.groupby('Department')['Salary'].mean().loc[lambda x:x > 50000]
```

```python
# Groupby + Count(frequency Analysis)
# -> Department me kitna employees 

df.groupby('Department')['Employee'].count()
```

```python
# GroupBY + Reset Index(Real DS Use)

result = df.groupby('Department')['Salary'].mean().reset_index()

print(result)
```

```python
# GroupBY on Multiply Columns 

# -> Department + Age group 

df.groupby(['Department', 'Age'])['Salary'].mean().reset_index()
```

```python
## Common Aggregation Function
# mean -> Average
# sum  -> Total
# count -> Frequency
# max ->  Highest
# min -> Lowest
# median -> Robust average 
```

```python
# Piviot Table 
# -> Data ko re-shape karke = group + aggregate + compare karna 

data = {
    "City": ["Delhi", "Delhi", "Mumbai", "Mumbai", 'Goa'],
    "Product": ["A", "B", "A", "B", 'A'],
    "Sales": [100, 150, 200, 180, 200]
}

df = pd.DataFrame(data)
```
