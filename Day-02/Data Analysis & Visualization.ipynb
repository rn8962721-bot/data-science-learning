## Data Understanding 

```python
##  Data Type 
# 1. Numerical data 
# -> Numbers jinpe Calculations hota hai 

# (a) Continuous 
#       - Float values
#       - Example: salary, temperature, revenue 

import pandas as pd 

df = pd.DataFrame({
    'salary': [45000.5, 52000.0, 61000.75 ]
})

df.dtypes

# Use case:
#   - Mean, median, std
#   - Regression models
#   - Line Plots 
```

```python
# (b) Discrete 
#       - integre counts 
#       - Example : number of purchases, clicks 


df['Purchase'] = [3, 1, 7]

# Use case:
#       - Count-based analysis 
#       - Bar charts
```

```python
# 2. Categorical data 
# ->  Labels / groups 

# (a) Nominal (No order)
# -> Gender, cily, product category  

df['City'] = ['Delhi', 'Mumbai', 'Delhi']

# Use case:
#   - Groupby
#   - One-hot encoding
#   - Bar plots 
```

```python
# (b) Ordinal (order matters)
#       - Low < Medium < High
#       - Rating(1-5)

df['rating'] = ['Low', 'High', 'Medium']

# Machine learning se pehle order preserve karna padta hai 
```

```python
# 3. Boolean Data 
#       - True / False
#       - Yes / No

df['is_active']  = [True, False, True]

# Use case:
#   - Filtering 
#   - Conditions 
#   - Binary classification 
```

```python
# 4. Data & Time Data (Very Important)
# Mostly row data string me hota hai

df['date'] = pd.to_datetime(["2024-01-01", "2024-01-05", "2024-01-10"])

# Use Case:
#   - Time series 
#   - Trend analysis
#   - Monthly / Yearly aggregation
```

```python
# 4. Pandas Series(1D data)
#       - Single column data 

s = pd.Series([100, 200, 300])

# Think: Excel ka single column 
```

```python
# 5. Pandas DataFrame (Most Important )
#       - 2D tabular data 
#       - 90% data science work yahin hota hai 

df = pd.DataFrame({
    "name": ["A", "B", "C"],
    "score": [80, 90, 85]
})

df.head()  # Excel sheet/ sql table
```

```python
### Data quality assessment 
#        - Rule#1: Bad data + best model = bat result 
# Data Quality Assessment ka matlab hai: 
#           - kya main is data pa barosa kar sakta hoon?

# 1. Dimensions of Data Quality
# (a) Completeness (Missing data)
# Question: 
#           - kya data poora hai?

df.isnull().sum()
```

```python
# (b) Correct Data Type 
# Question
#       - kay data sahi type me hai 

df.dtypes

# commong issues:
#      - Numbers as object
#      - Dates as string 

df['date'] = pd.to_datetime(df['date'], errors='coerce')
df['price'] = pd.to_numeric(df['price'], errors='coerce')
```

```python
# (c)  Accuracy (Logical correctness)
#  Question: 
#           - Data logically prossible hai?
# Example:
#   - Age = -5
#   - salary = 1 crore for inten

df[df['age'] < 0]
df[df['salary'] > 100000] 

# Business logic lagao , sife code nahi.
```

```python
# (d) Consistency
# Question:
#       - Same cheez multiple formats me toh nahi?

df["city"].value_counts()

# Examples:
# Delhi, delhi, DELHI
# Yes, Y, true

df["city"] = df["city"].str.lower()
```

```python
# (e) Uniqueness (Duplicates)
# Question:
#   - Duplicate rows / IDs toh nahi?

df.duplicated().sum()

df.drop_duplicates(inplace=True)

#Important:
# Primary key duplicates = data issue, not pandas issue.
```

```python
# (f) Valid Range / Outliers
# Question:
# Data expected range me hai?

df.describe()

df[df["salary"] > df["salary"].quantile(0.99)]

# Note:
#  Outlier ≠ always wrong
#  Sometimes important signal hota hai.
```

```python
# Quick Data Quality Checklist (Interview + Project)

df.shape          # rows, columns
df.info()         # types + missing
df.describe()     # stats
df.head()         # sample check
df.nunique()      # categorical sanity

# gar yeh 5 cheezein kar li →
# 70% data quality issues pakad liye
```
